```python
# Calculating the percentage of missing values in the DataFrame
dti_missing = applications['dti'].isna().sum()
total = len(applications)

dti_missing_percentage = dti_missing / total * 100
print(f'Percentage of missing values in the dti column: {dti_missing_percentage:.2f}%')
```

```python
# Calculating the percentage of missing values in rejection_reason column
rejection_reason_missing = applications['rejection_reason'].isna().sum()
rejection_reason_missing_percentage = rejection_reason_missing / total * 100
print(f'Percentage of missing values in the rejection_reason column: {rejection_reason_missing_percentage:.2f}%')
```

```python
# know how many rejection reasons exist for each application
applications.groupby("application_status")["rejection_reason"].apply(
    lambda x: x.isna().sum()
)
```

```python
# Checking the Missing values in the Loans DataFrame
print('Missing values in the loans DataFrame:')
print(loans.isnull().sum())
```

```python
# Checking the duplicate values in the Primary key columns of the DataFrames
print('Duplicate values in the Primary key in the  DataFrame:')
print("Customer ID duplicates:", customers["customer_id"].duplicated().sum())

print("Application ID duplicates:", applications["application_id"].duplicated().sum())

print("Loan ID duplicates:", loans["loan_id"].duplicated().sum())
```

```python
# Checking the relationship between the three DataFrames using the primary keys
print('The relationship between the three DataFrames: ')
invalid_customer_ids = applications[~applications['customer_id'].isin(customers['customer_id'])]
print(f'Number of invalid customer IDs in applications DataFrame: {len(invalid_customer_ids)}') 

invalid_application_ids = loans[~loans['application_id'].isin(applications['application_id'])]
print(f'Number of invalid application IDs in loans DataFrame: {len(invalid_application_ids)}')
```

```python
# Checking the applucation status distribution in the applications DataFrame
application_status_counts = applications['application_status'].value_counts(normalize=True) * 100
print("Application Status Distribution:")
print(application_status_counts)
```

```python
# calculating the approval rate for the applications DataFrame
approval_rate = applications['application_status'] == 'Approved'
approval_rate = approval_rate.mean() * 100
print(f'Approval Rate: {approval_rate:.2f}%')
```

```python
# checking the critcal business rule
loan_application_check = loans.merge(
    applications[["application_id", "application_status"]],
    on="application_id",
    how="left"
)
rejected_loans = loan_application_check[
    loan_application_check["application_status"] == "Rejected"
]
print(f'Number of loans associated with rejected applications: {len(rejected_loans)}')
```

```python
pending_loans = loan_application_check[
    loan_application_check["application_status"] == "Pending"       
]

print(f'Number of loans associated with pending applications: {len(pending_loans)}')
```

```python
# checking the Loan Amounts
print("Minimum loan:", loans["approved_amount"].min())
print("Maximum loan:", loans["approved_amount"].max())
print("Average loan:", loans["approved_amount"].mean())
```

```python
invalid_amounts = loans[loans["approved_amount"] <= 0]

print("Invalid loan amounts:", len(invalid_amounts))
```
